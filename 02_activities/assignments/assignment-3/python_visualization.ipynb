```python
# Import Libraries 
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import scipy
import PIL
import requests
```

```python
# Import data 
data = pd.read_csv("CafeTOLocations.csv")
#print(data)

# Group data by Ward and intervention type

cafe_types = data.groupby(['WARD_NAME', 'INTERVENTION_TYPE']).size().reset_index(name='Count')

print(cafe_types)
```

```python
# Set plot style
sns.set(style="whitegrid")
# Create grouped bar plot
plt.figure(figsize=(20, 10))
ax = sns.barplot(x='WARD_NAME', y='Count', hue='INTERVENTION_TYPE', data=cafe_types, palette='colorblind')

# Chart labels and formatting
plt.title('Number of CaféTO Interventions by Ward and Type', fontsize=20)
plt.xlabel('Ward', fontsize=16)
plt.ylabel('Number of Cafés', fontsize=14)
plt.xticks(rotation=45)
plt.legend(title='Intervention Type')
plt.tight_layout()

# Edit legend labels to get rid of the the weird 'É' thing from the csv file
# Used code from stackoverflow for this!
handles, labels = ax.get_legend_handles_labels()
new_labels = ['Private Property Patio', 'Sidewalk Café', 'Curb Lane Café']
ax.legend(handles=handles, labels=new_labels, title='Café Type')
# Show plot
plt.show()
```
